```python
import numpy as np
```

```python
# NUMPY FILE HANDLING
# -------------------

'''
NumPy provides simple functions to save and load arrays.

The main file formats are:

.npy  → one NumPy array
.npz  → multiple NumPy arrays
'''


'''
1. ONE ARRAY — .npy
-------------------

Array
  ↓
np.save()
  ↓
.npy file
  ↓
np.load()
  ↓
NumPy array
'''

# Create a NumPy array.
array1 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# Save one array to a .npy file.
np.save("array1.npy", array1)

# Load the array back from the file.
loaded_array1 = np.load("array1.npy")

print("Loaded array:")
print(loaded_array1)

# Check the loaded array.
print("Type:", type(loaded_array1))
print("Shape:", loaded_array1.shape)


'''
2. MULTIPLE ARRAYS — .npz
--------------------------

Multiple arrays
      ↓
  np.savez()
      ↓
   .npz file
      ↓
   np.load()
      ↓
Multiple arrays
'''

# Create multiple arrays.
array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])
array3 = np.array([7, 8, 9])

# Save multiple arrays into one .npz file.
# Give each array a name so we can access it later.
np.savez(
    "arrays.npz",
    first=array1,
    second=array2,
    third=array3
)

# Load the .npz file.
loaded = np.load("arrays.npz")

# Check which arrays are stored inside the file.
print("Available arrays:", loaded.files)

# Access arrays using their names.
print("First array:", loaded["first"])
print("Second array:", loaded["second"])
print("Third array:", loaded["third"])


'''
3. COMPRESSED ARRAYS
--------------------

Multiple arrays
      ↓
np.savez_compressed()
      ↓
Compressed .npz file
      ↓
    np.load()
      ↓
Multiple arrays
'''

# Save multiple arrays with compression.
# Compression can reduce the file size.
np.savez_compressed(
    "arrays_compressed.npz",
    first=array1,
    second=array2,
    third=array3
)

# Load the compressed .npz file normally.
loaded_compressed = np.load("arrays_compressed.npz")

print("Available arrays:", loaded_compressed.files)

print("First array:", loaded_compressed["first"])
print("Second array:", loaded_compressed["second"])
print("Third array:", loaded_compressed["third"])


'''
IMPORTANT
---------

.npy
→ Stores one NumPy array.

.npz
→ Stores multiple NumPy arrays.

np.save()
→ Save one array.

np.load()
→ Load an array or .npz file.

np.savez()
→ Save multiple arrays in one .npz file.

np.savez_compressed()
→ Save multiple arrays in a compressed .npz file.

loaded.files
→ Shows the names of arrays stored inside a .npz file.


WHEN TO USE
-----------

One array
→ np.save()

Multiple arrays
→ np.savez()

Multiple arrays + smaller storage
→ np.savez_compressed()


REAL-WORLD USE
--------------

NumPy file handling is useful for:

• Saving processed numerical data
• Reusing arrays later
• Storing ML features
• Saving preprocessed datasets
• Avoiding expensive recalculation
• Saving intermediate results in ML workflows
'''
```
