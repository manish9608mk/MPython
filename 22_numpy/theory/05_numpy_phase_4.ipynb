```python
import numpy as np
import matplotlib.pyplot as plt
```

##### NUMPY FILE HANDLING
1. Create arrays
2. Save arrays using np.save()
3. Load arrays using np.load()

```python
# Create a 2D array.
array1 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# Create a 3 × 3 array with random values between 0 and 1.
array2 = np.random.rand(3, 3)

# Create a 4 × 4 array filled with zeros.
array3 = np.zeros((4, 4))
```

```python
# Save arrays using np.save()
# Save each NumPy array into a separate .npy file.
# .npy → stores one NumPy array.

np.save("array1.npy", array1)
np.save("array2.npy", array2)
np.save("array3.npy", array3)

'''
think:
array1 → np.save() → array1.npy
array2 → np.save() → array2.npy
array3 → np.save() → array3.npy
'''
```

```python
# Load arrays using np.load()
# Load the saved arrays back into NumPy.
loaded_array1 = np.load("array1.npy")
loaded_array2 = np.load("array2.npy")
loaded_array3 = np.load("array3.npy")

print("Array 1:")
print(loaded_array1)

print("\nArray 2:")
print(loaded_array2)

print("\nArray 3:")
print(loaded_array3)
```

```python
# Check the basic properties of the loaded array.
print("Shape:", loaded_array1.shape)
print("Data type:", loaded_array1.dtype)
print("Number of dimensions:", loaded_array1.ndim)
print("Total elements:", loaded_array1.size)

# Check the loaded array
# This is useful in real projects because you often want to verify what you loaded.
```

```python
# Save multiple arrays in one file
# .npz → stores multiple NumPy arrays in one file.
np.savez(
    "restaurant_data.npz",
    sales=array1,
    random_data=array2,
    empty_data=array3
)

# load it
# Load the .npz file.
data = np.load("restaurant_data.npz")

# Check which arrays are stored inside.
print("Stored arrays:", data.files)

# Access arrays using their names.
print("\nSales:")
print(data["sales"])

print("\nRandom data:")
print(data["random_data"])

print("\nEmpty data:")
print(data["empty_data"])


'''
.npy
 ↓
One NumPy array

.npz
 ↓
Multiple NumPy arrays
'''
```

```python
# Compressed .npz
# When arrays are large, compression can reduce file size.

# Save multiple arrays with compression.
np.savez_compressed(
    "restaurant_data_compressed.npz",
    sales=array1,
    random_data=array2,
    empty_data=array3
)

# Load normally:
# Load the compressed file.
compressed_data = np.load("restaurant_data_compressed.npz")

print(compressed_data.files)
print(compressed_data["sales"])
```

```python
# Check whether the saved file exists
# Useful when working with datasets or ML pipelines.

from pathlib import Path

file_path = Path("array1.npy")

if file_path.exists():
    print("File exists.")
else:
    print("File does not exist.")
```

### LOAD AND DISPLAY NUMPY LOGO

```python
try:
    # Load the saved NumPy array.
    logo = np.load("numpy-logo.npy")

    # Check what we loaded.
    print("Shape:", logo.shape)
    print("Dimensions:", logo.ndim)
    print("Data type:", logo.dtype)

    # Create one figure with 2 plots.
    plt.figure(figsize=(10, 5))

    # ORIGINAL LOGO
    plt.subplot(121)
    plt.imshow(logo)
    plt.title("NumPy Logo")
    plt.axis("off")

    # DARK LOGO
    # Change the pixel values to create an inverted version.
    dark_logo = 1 - logo

    plt.subplot(122)
    plt.imshow(dark_logo)
    plt.title("NumPy Dark Logo")
    plt.axis("off")

    # Display both plots together.
    plt.show()

except FileNotFoundError:
    print("NumPy logo file not found.")
```
