## Advance operation with business example

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#### Data structure: [restaurant_id, 2021, 2022, 2023, 2024]

sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print("------- Zomato sales analysis -------")

# SHAPE
# 5 restaurants × 5 columns
print("Shape:", sales_data.shape)

# FIRST 3 RESTAURANTS
print("\nFirst 3 restaurants:")
print(sales_data[:3])

# RESTAURANT IDs
print("\nRestaurant IDs:")
print(sales_data[:, 0])

# 2021 SALES
print("\n2021 sales:")
print(sales_data[:, 1])

# 2024 SALES
print("\n2024 sales:")
print(sales_data[:, 4])

# TOTAL SALES IN 2024
print("\nTotal sales in 2024:")
print(sales_data[:, 4].sum())

# AVERAGE SALES IN 2024
print("\nAverage sales in 2024:")
print(sales_data[:, 4].mean())

# HIGHEST SALES IN 2024
print("\nHighest sales in 2024:")
print(sales_data[:, 4].max())

# LOWEST SALES IN 2024
print("\nLowest sales in 2024:")
print(sales_data[:, 4].min())


# filtering
# Restaurants with 2024 sales greater than 250000
high_sales = sales_data[sales_data[:, 4] > 250000]

print("\nRestaurants with 2024 sales > 250000:")
print(high_sales)


# argmax()
# Find the index of the restaurant with highest 2024 sales
index = np.argmax(sales_data[:, 4])

print("\nIndex of highest 2024 sales:", index)
print("Restaurant data:", sales_data[index])
```

```python
# total sale per year 
print(np.sum(sales_data, axis=0))

# or,
yearly_total = np.sum(sales_data[:, 1:], axis=0)
print(yearly_total)
```

```python
# minimum sales per restaurant 2021
min_sales = np.min(sales_data[:, 1:], axis=1)
print(min_sales)
```

```python
# avarage sales per restaurant 2021
avg_sales = np.mean(sales_data[:, 1:], axis=1)
print(avg_sales)
```

```python
# maximum sales per year 2021
max_sales = np.max(sales_data[:, 1:], axis=0)
print(max_sales)
```

```python
# CUMULATIVE SUM OF SALES

'''
Understand cumsum() simply

Suppose one restaurant has:

2021 → 150000
2022 → 180000
2023 → 220000
2024 → 250000

np.cumsum() gives:

2021 → 150000
2022 → 330000     # 150000 + 180000
2023 → 550000     # 330000 + 220000
2024 → 800000     # 550000 + 250000
'''

# Add sales year by year for each restaurant.
cum_sum = np.cumsum(sales_data[:, 1:], axis=1)
print(cum_sum)



# MATPLOTLIB

# Calculate average cumulative sales of all restaurants for each year.
average_cumulative_sales = np.mean(cum_sum, axis=0)

# Years for the X-axis.
years = [2021, 2022, 2023, 2024]

# Create the graph.
plt.figure(figsize=(5, 4))

# Plot average cumulative sales against years.
plt.plot(years, average_cumulative_sales, marker="o")

plt.title("Average Cumulative Sales Across All Restaurants")
plt.xlabel("Years")
plt.ylabel("Sales")

# Show only the actual years on the X-axis.
plt.xticks(years)
plt.grid(True)

# Display the graph.
plt.show()
```

##### VECTOR OPERATIONS

1. Vector addition
2. Vector multiplication
3. Dot product
4. Magnitude / Norm
5. Angle between vectors
6. Normalization
7. Euclidean distance
8. Cross product → basic understanding

```python
vector1 = np.array([1, 2, 3, 4, 5])
vector2 = np.array([6, 7, 8, 9, 10])


# 1. VECTOR ADDITION
# Add corresponding elements.
print("Vector addition:")
print(vector1 + vector2)


# 2. VECTOR MULTIPLICATION
# Multiply corresponding elements.
print("\nVector multiplication:")
print(vector1 * vector2)


# 3. DOT PRODUCT
# Multiply corresponding elements and add them.
print("\nVector dot product:")
print(np.dot(vector1, vector2))


# 4. MAGNITUDE / NORM
# Magnitude = length of a vector.
print("\nVector 1 magnitude:")
print(np.linalg.norm(vector1))

print("Vector 2 magnitude:")
print(np.linalg.norm(vector2))


# 5. ANGLE BETWEEN VECTORS
# Formula:
# cos(θ) = (A · B) / (|A| × |B|)

cos_angle = np.dot(vector1, vector2) / (
    np.linalg.norm(vector1) * np.linalg.norm(vector2)
)

angle = np.arccos(cos_angle)

print("\nAngle in radians:")
print(angle)

print("Angle in degrees:")
print(np.degrees(angle))


# 6. NORMALIZATION
# Convert a vector into a unit vector.
# A unit vector has magnitude = 1.

normalized_vector1 = vector1 / np.linalg.norm(vector1)

print("\nNormalized vector:")
print(normalized_vector1)

print("Normalized vector magnitude:")
print(np.linalg.norm(normalized_vector1))


# 7. EUCLIDEAN DISTANCE
# Measures the straight-line distance between two vectors.

distance = np.linalg.norm(vector1 - vector2)

print("\nEuclidean distance:")
print(distance)


# 8. CROSS PRODUCT
# Returns a vector perpendicular to both vectors.
# Cross product is defined for 3D vectors.

cross_product = np.cross(vector1[:3], vector2[:3])

print("\nCross product:")
print(cross_product)
```

##### VECTORIZED OPERATION
* np.vectorize() → useful when you want to apply a Python function element-by-element to an array.

```python
restaurant_types = np.array([
    'biryani',
    'chinese',
    'pizza',
    'burger',
    'cafe'
])

vectorized_upper = np.vectorize(str.upper)

print("Vectorized Upper:", vectorized_upper(restaurant_types))
```

##### BROADCASTING
* Broadcasting = NumPy automatically adjusts compatible array shapes so that operations can be performed without manually reshaping or looping.

```python
# BROADCASTING WITH SCALAR

# Take only yearly sales.
yearly_sales = sales_data[:, 1:]

# Divide every sales value by 12.
# NumPy broadcasts 12 across the entire array.
monthly_avg = yearly_sales / 12

print("Monthly average sales:")
print(monthly_avg)
```
