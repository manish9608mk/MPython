# Why These Tools Matter

Use the tool for the job.

```text
Application code
    → VS Code / PyCharm / Cursor

Exploration / Data
    → Jupyter

Interactive experimentation
    → IPython

Cloud notebook
    → Colab / Kaggle

Environment management
    → venv / Conda
```

The goal is not collecting tools. The goal is knowing
why and when to use each one.

# Practical Goal

A clean Python project should eventually have:

```text
source code
+
tests
+
formatter / linter
+
type hints
+
requirements / project metadata
+
README
+
.gitignore
```

This is more important than learning another IDE.
