# 00 — MPython Overview

This folder contains focused notebooks for the major areas
listed in the main `README.md`.

Each notebook answers two questions:

1. What have I studied already?
2. What should I learn next?

This is intentionally split into multiple files because a
single giant learning document becomes difficult to use and update.
