# Target Cloud Project

Build and deploy one real backend application:

```text
Client
  ↓
Load Balancer / API Gateway
  ↓
FastAPI
  ↓
PostgreSQL / RDS

S3
 ↓
Files / artifacts

IAM
 ↓
Workload permissions

CloudWatch
 ↓
Logs + metrics
```

Then automate infrastructure and deployment using:

```text
Terraform
+
GitHub Actions
+
Docker
+
ECR
+
ECS / Fargate
+
CloudWatch
```

The goal is not to memorize AWS services. The goal is to
understand how they fit together to run an application.
