# Standard Workflow

```text
Create project
    ↓
Create `.venv`
    ↓
Activate environment
    ↓
Install dependencies
    ↓
Develop
    ↓
Freeze dependencies
    ↓
requirements.txt
```

## Important Principle

A project should keep its dependencies isolated from other
projects so upgrades do not unexpectedly break unrelated work.


# Practical Goal

Take one existing Python project and make it cleanly reproducible:

```text
clone
  ↓
create environment
  ↓
install dependencies
  ↓
run application
  ↓
run tests
```

If another machine can reproduce the project reliably,
the environment setup is doing its job.
