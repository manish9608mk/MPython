# NumPy Array Functions

NumPy provides many built-in functions for working with arrays.

These functions can be used to:

- find unique values
- perform mathematical operations
- round numbers
- compare values
- find positions of values
- sort arrays
- manipulate arrays

Commonly used NumPy array functions include:

- `np.unique()`
- `np.sqrt()`
- `np.abs()`
- `np.power()`
- `np.exp()`
- `np.log()`
- `np.round()`
- `np.floor()`
- `np.ceil()`
- `np.clip()`
- `np.where()`
- `np.argmax()`
- `np.argmin()`
- `np.argsort()`
- `np.sort()`
- `np.sum()`
- `np.mean()`
- `np.min()`
- `np.max()`

Many NumPy functions work element-wise, meaning the operation is applied independently to each element of an array.

```python
numbers = np.array([150, 200, 150, 300, 250, 200, 350, -50])

# Unique values
print("Unique:", np.unique(numbers))

# Square root
positive_numbers = np.array([4, 9, 16, 25])
print("Square Root:", np.sqrt(positive_numbers))

# Absolute values
print("Absolute:", np.abs(numbers))

# Power
print("Squared:", np.power(numbers, 2))

# Rounding
decimal_numbers = np.array([1.234, 2.567, 3.891])

print("Rounded:", np.round(decimal_numbers, 2))
print("Floor:", np.floor(decimal_numbers))
print("Ceil:", np.ceil(decimal_numbers))

# Clip values to a range
scores = np.array([20, 50, 80, 105, 120])
print("Clipped Scores:", np.clip(scores, 0, 100))

# Maximum and its index
print("Maximum:", np.max(numbers))
print("Maximum Index:", np.argmax(numbers))

# Minimum and its index
print("Minimum:", np.min(numbers))
print("Minimum Index:", np.argmin(numbers))

# Sorting
print("Sorted:", np.sort(numbers))
print("Sorting Indices:", np.argsort(numbers))

# Conditional selection
print("Pass/Fail:", np.where(scores >= 50, "Pass", "Fail"))

# Exponential
print("Exponential:", np.exp([1, 2, 3]))

# Natural logarithm
print("Logarithm:", np.log([1, np.e, np.e**2]))
```
