# Student Performance Analysis using NumPy
------------------------------------------

## Problem Statement
--------------------

Analyze student performance using NumPy.

The dataset contains:
- Study hours
- Attendance
- Assignment scores
- Exam scores

The goal is to analyze the data, identify students who need attention, find top students, calculate final scores, and visualize performance.

## Analysis Performed
---------------------

- Generated student data using NumPy.
- Explored array size and shape.
- Calculated basic statistics.
- Identified students with low attendance.
- Found top 5 students using `np.argsort()`.
- Performed pass/fail analysis.
- Calculated weighted final scores.
- Found the highest-scoring student using `np.argmax()`.
- Visualized the data using Matplotlib.

## Tools
---------

Python • NumPy • Matplotlib • Jupyter Notebook

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Make random results reproducible.
np.random.seed(42)

# Student IDs
student_ids = np.arange(1, 21)

# Study hours per week
study_hours = np.random.randint(5, 31, 20)

# Attendance percentage
attendance = np.random.randint(60, 101, 20)

# Assignment scores out of 100
assignment_scores = np.random.randint(40, 101, 20)

# Exam scores out of 100
exam_scores = np.random.randint(35, 101, 20)

print("Student IDs:", student_ids)
print("Study Hours:", study_hours)
print("Attendance:", attendance)
print("Assignment Scores:", assignment_scores)
print("Exam Scores:", exam_scores)
```

##### 2. Understand the data

```python
print("Number of students:", student_ids.size)

print("Study hours shape:", study_hours.shape)
print("Attendance shape:", attendance.shape)
print("Assignment scores shape:", assignment_scores.shape)
print("Exam scores shape:", exam_scores.shape)
```

##### 3. Basic statistics

```python
print("Average study hours:", np.mean(study_hours))
print("Maximum study hours:", np.max(study_hours))
print("Minimum study hours:", np.min(study_hours))

print("\nAverage attendance:", np.mean(attendance))
print("Average assignment score:", np.mean(assignment_scores))
print("Average exam score:", np.mean(exam_scores))
```

##### 4. Find students who need attention

```python
# Students with attendance below 75%.
low_attendance = student_ids[attendance < 75]

print("Students with low attendance:")
print(low_attendance)
```

##### 5. Find top students

```python
# Get the indexes that would sort exam scores.
ranking = np.argsort(exam_scores)[::-1]

# Top 5 students.
top_5 = ranking[:5]

print("Top 5 students:")

for index in top_5:
    print(
        "Student:", student_ids[index],
        "| Exam Score:", exam_scores[index]
    )
```

##### 6. Pass / Fail

```python
# Let's say passing marks are 40.
# True where the student passed.
passed = exam_scores >= 40

print("Pass/Fail:")
print(passed)

print("\nNumber of students passed:", np.sum(passed))
print("Number of students failed:", np.sum(~passed))

pass_percentage = np.mean(passed) * 100
print("Pass percentage:", pass_percentage, "%")
```

##### 7. Create a final score

```python
final_score = (
    assignment_scores * 0.3
    + exam_scores * 0.7
)

print("Final scores:")
print(final_score)

# find the best overall student
best_student_index = np.argmax(final_score)

print("Best student:", student_ids[best_student_index])
print("Final score:", final_score[best_student_index])
```

##### 8. Visualization

```python
plt.figure(figsize=(8, 5))

plt.scatter(study_hours, exam_scores)

plt.xlabel("Study Hours per Week")
plt.ylabel("Exam Score")
plt.title("Study Hours vs Exam Score")

plt.grid(True)
plt.show()
```

##### exam-score distribution

```python
plt.figure(figsize=(8, 5))

plt.hist(exam_scores, bins=5)

plt.xlabel("Exam Score")
plt.ylabel("Number of Students")
plt.title("Exam Score Distribution")

plt.show()
```
